## Problem 1: String alphabet analysis [13 points]


In the cell below, write a piece of code that will do the following work:

* (1) Introduce an arbitrary string of English text (Suppose this is text from a book, or an article on the web, etc.) The string should be of length 300, and contain some numbers, such as 1 or 27 etc, and may also contain . Have your code to print the length of the string, and the number of words in the string, and have your program print the results in a clear way. For example, the string `"The weather today is 2 times better than yesterday; it is not raining. Excellent!"` contains 14 words. [5 points] 

* (2) Print the string in reverse, also replacing all instances of the letter 'b' with the letter 'X', and printing everything in the lower case. [2]

* (3) Write code to calculate the number of times the letter 'a' appears in the string, and print the result. Use the `format` method in the print statement. Then, using the python `for` loop, do the same for the remaining letters in the alphabet, counting lower case letters. Also create a list of all the letters missing from your string and print it. [4 points]

* (4) Remove all digits between 0 to 8 from the string. Also, find any digits `9` in the string and replace them with the word 'nine'. [2 points]


Hints:

````python
#Your Answer to Problem 1 in this cell:
# (1)
import string
text = """In 2024, a team of 15 archaeologists found a 3000-year-old tomb hidden 45 meters beneath the sands of Egypt.
Inside, they found 9 large chambers containing 142 gold artifacts and 27 unique pottery jars.
The expedition lasted 18 days, costing exactly 50,000 dollars; all their findings were priceless!"""
print("="*120)
print(text)
print("="*120)
# split on whitespace
print("""1.1: The above string contains {0} characters and {1} words.""".format(len(text),len(text.split())))
print()
print("1.2: String in reverse, 'b' replaced by 'X' then lower case :")
print(text[::-1].replace('b','X').lower())
print()
print("""1.3: Total times 'a' appear in string: {}.""".format(text.count('a')))
print()
print('Other lower case letter appearance in string:\n')
# storing value in dict
count = {}
for i in string.ascii_lowercase:
    if i != 'a':
        count[i] = text.count(i)

# Extra: Sorting frequency by value for easy visual
count = dict(sorted(count.items(),key=lambda x:x[1],reverse=True))
for k,v in count.items():
    print("""Total times '{0}' appear in string: {1}.""".format(k,v))
print('\nall the letters missing from string:\n')
print([i for i in string.ascii_lowercase if i not in text])
print()
print('1.4: Removed all 0 to 8 and replace `9` with `nine`')
print()
print(''.join(i.replace('9','nine') for i in text if i not in '012345678'))
````

### Justification:
**1.1 :** 
- `len` returns numbers of items in a container which includes objects like characters, whitespace, new line. As the text contains all of these in total it returns 300 characters in the text. The `format` method is used to inserts these values with texts in `print` statement.

- `split` without any argument (split by whitespace including new lines) used to generate words in a list. Therefore, `len(text.split())` gives total words count as 48.


**1.2 :**  After reversing the text by slicing(`[::-1]`), replaced `'b'` with `'X'` with `replace` then printed everyting in lower case.

**1.3 :**  

- Firstly, `count` is used to count the occurances of `'a'` in string.
- Then, a for loop is used to count other lower letters in similar way.
- Extra (not required by the task): Used a dict to store the letter-count pair as key-value pair, sorted by value so that `print` shows highest frequency letters as first.
- For missing value used a list comprehension method with condition on iterators to print the lower case letter not present in text.

**1.4 :**  Used a generator with `join` to remove all occurance of `'0 to 8'` except `'9'` which is replace by text `'nine'`.

## Problem 2. Dictionary and breaking the loop [12 points]

Introduce a string `s` (e.g., a sentence) with 10 different words in it. 

* Create a dictionary called `d` where the key is the order of the word in the string (remeber, we start with index 0 in python), and the value is the word. Print `d`. [3]

* Run a `for` loop and draw 20 random numbers from 0 to 9. These numbers are our random keys. Using the dictionary `d`, create a new sentence in a string `s_new` by adding the words corresponding to the random keys in `d`. Print `s_new`. [4]

* Now use another `for` loop to go over the dictionary `d`. For each word in `d`, count the number of times it is present in `s_new`. If the loop encounters a word that is in `d` but not in `s_new`, then exit the loop with the `break` statement, printing the message  `'word' is missing in s_new`, where `word` is the missing one. [5]

````python
#Your Answer to problem 2 in this cell:
s = "A quick brown fox jumps over the lazy dog today"
print("Sentence with {0} differnt words".format(len(s.split())))
print("="*50)
print(s)
print("="*50)
print("2.1: Dict of words with index") 
# dictionary
d = {}
# split on whitespace
word = s.split()
for index in range(len(word)):
    d[index]=word[index]
print(d)
print()
print("2.2 New sentnce with random suffle of words:") 
import random
# new string
s_new =''
for i in range(20):
     random_keys = random.randint(0,9)
    # adding values with a space
     s_new += d[random_keys]+' '
print(s_new)
print()
print("2.3:") 
for word in d.values(): 
    if word not in s_new:
        print("'{0}' is missing in s_new".format(word))
        break
    print("Number of times '{0}' present in s_new: {1}".format(word,s_new.count(word)))
````

### Justification:
**2.1 :** Splitting the words by whitespace creates a list of words. The index of each word in the list used as dictionary key, and the word itself as a value.

**2.2 :** The randomly generated values are used as key to access the dictionary values. These values then iteratively appended to generate a new string, A space string with each word is added to generate new sentence.

**2.3 :** The loops checks if any word present in dictionary (as value) is also present in new sentence or not. If not, the loop exists with `break`. Otherwise, it printed how many times each word appears in the new sentence with `count` which return number of times a non-overlapping substring present in a sting.
